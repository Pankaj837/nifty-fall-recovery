```python
import yfinance as yf
import pandas as pd

df = yf.download("^NSEI", start="2007-01-01", auto_adjust=False)
df.columns = df.columns.get_level_values(0)
df = df.reset_index()
df.to_csv("../data/raw_nifty.csv", index=False)

print(df.head())
print(df.tail())
print(df.shape)
print(df.dtypes)
```

Source: Yahoo Finance via yfinance, ticker ^NSEI (NIFTY 50)
Downloaded: 2026-09-23
Range: 2007-09-17 to 2026-09-23 (4665 rows)
Fields: Date, Open, High, Low, Close, Adj Close, Volume
Note: Volume is 0 for the index in many rows and is not used. Raw file is never edited.
